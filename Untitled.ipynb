```python
class GroupStdFeatures():
    def __init__(self, group_col='card1', value_col='TransactionAmt', fallback_to_global=False):
        self.group_col = group_col
        self.value_col = value_col
        self.fallback_to_global = fallback_to_global

    def fit(self, X, y=None):
        stats = X.groupby(self.group_col)[self.value_col].agg(['mean', 'std'])
        self.mean_ = stats['mean']
        self.std_ = stats['std']
        self.global_mean_ = X[self.value_col].mean()
        self.global_std_ = X[self.value_col].std()
        return self

    def transform(self, X):
        X = X.copy()
        mean = X[self.group_col].map(self.mean_)
        std = X[self.group_col].map(self.std_)

        if self.fallback_to_global:
            mean = mean.fillna(self.global_mean_)
            std = std.fillna(self.global_std_)

        std = std.replace(0, np.nan)   # нулевой std даст inf

        X[f'{self.value_col}_std_{self.group_col}'] = std
        X[f'z_score_trAmnt'] = (X[self.value_col] - mean) / std
        X['trAmnt_div_std'] = X[self.value_col]/std

        return X

transformer = GroupStdFeatures().fit(X_train)
X_train_stats, X_val_stats = transformer.transform(X_train), transformer.transform(X_val)

train_pool_stats = Pool(
    data=X_train_stats,
    label=y_train,
    cat_features=all_cat_columns
)

val_pool_stats = Pool(
    data=X_val_stats,
    label=y_val,
    cat_features=all_cat_columns
)
model2 = CatBoostClassifier(
    iterations=1500,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU'
)
model2.fit(train_pool_stats, eval_set=val_pool_stats)
```

```python
def group_columns(data,id_30 = True, id_31 = True, id_33 = True,):
    data= data.copy()
    if id_30:
        conditions = [
        data['id_30'].str.contains('Windows', case=False, na=False),
        data['id_30'].str.contains('iOS', case=False, na=False),
        data['id_30'].str.contains('Android', case=False, na=False),
        data['id_30'].str.contains('Mac', case=False, na=False),
        data['id_30'].str.contains('Linux', case=False, na=False),
        ]
        choices = ['Windows', 'iOS', 'Android', 'Mac', 'Linux']
        data['os_group'] = np.select(conditions, choices, default='Other')
        data.drop('id_30',axis= 1, inplace = True)
    if id_31:
        br_conditions = [
        data['id_31'].str.contains('chrome', case=False, na=False),
        data['id_31'].str.contains('safari', case=False, na=False),
        data['id_31'].str.contains('firefox', case=False, na=False),
        data['id_31'].str.contains('edge', case=False, na=False),
        data['id_31'].str.contains(r'\bie\b|internet explorer', case=False, na=False, regex=True),
        data['id_31'].str.contains('samsung', case=False, na=False),
        data['id_31'].str.contains('opera', case=False, na=False),
        ]
        choices = ['Chrome', 'Safari', 'Firefox', 'Edge', 'IE', 'Samsung', 'Opera']
        data['browser_group'] = np.select(br_conditions, choices, default='Other')
        data.drop('id_31', axis = 1,inplace = True)
    if id_33:
        data['id_33'] = np.where(
        data['id_33'].isna(),
        'missing',      # если NaN
        'present'       # если не NaN
        )
    return data

X_train_groups_stats, X_val_groups_stats = group_columns(X_train_stats), group_columns(X_val_stats)
gr_cat_columns = X_train_groups_stats.select_dtypes(include=['object', 'category', 'string']).columns.tolist()

train_group_pool = Pool(
    data=X_train_groups_stats,
    label=y_train,
    cat_features=gr_cat_columns
)
val_group_pool = Pool(
    data=X_val_groups_stats,
    label=y_val,
    cat_features=gr_cat_columns
)

model3 = CatBoostClassifier(
    iterations=1500,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU'
)
model3.fit(train_group_pool, eval_set=val_group_pool)
```

```python
import optuna
from sklearn.metrics import average_precision_score

def objective_catboost(trial):
    pos_weight = trial.suggest_float('pos_weight', 1.0, 20.0)
    params = {
        'iterations': trial.suggest_int('iterations', 500, 2000),
        'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.3, log=True),
        'depth': trial.suggest_int('depth', 4, 10),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 1e-1, 10.0, log=True),
        'random_strength': trial.suggest_float('random_strength', 1e-9, 10.0, log=True),
        'eval_metric': 'PRAUC',  # CatBoost оптимизирует PR AUC напрямую
        'verbose': 0,
        'random_seed': 42,
        'task_type':'GPU',
        'class_weights':[1.0, pos_weight]
    
    }
    
    model = CatBoostClassifier(**params)
    model.fit(train_pool_stats, eval_set=val_pool_stats, early_stopping_rounds=50, verbose=False)
    
    preds = model.predict_proba(val_pool_stats)[:, 1]
    return average_precision_score(y_val, preds)  # PR AUC из sklearn

# Запуск поиска CatBoost
study_cb = optuna.create_study(direction='maximize')
study_cb.optimize(objective_catboost, n_trials=30)

print("Best CatBoost Params:", study_cb.best_params)
print("Best CatBoost PR AUC:", study_cb.best_value)
```

```python
class CategoricalFiller(BaseEstimator, TransformerMixin):
    def __init__(self, columns=None, fill_value='missing'):
        self.columns = columns
        self.fill_value = fill_value

    def fit(self, X, y=None):
        # если колонки не заданы, берём все строковые/категориальные колонки из train
        if self.columns is None:
            self.columns_ = X.select_dtypes(include=['object', 'category', 'string']).columns.tolist()
        else:
            self.columns_ = list(self.columns)
        return self

    def transform(self, X):
        X = X.copy()
        for c in self.columns_:
            if c in X.columns:
                X[c] = X[c].astype(object).where(X[c].notna(), self.fill_value).astype(str)
        return X

    # def fit_transform(self,X,y= None):
    #     temp = self.fit(X,y)
    #     return self.transform(temp)

class GroupStdFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, group_col='card1', value_col='TransactionAmt', fallback_to_global=False):
        self.group_col = group_col
        self.value_col = value_col
        self.fallback_to_global = fallback_to_global

    def fit(self, X, y=None):
        stats = X.groupby(self.group_col)[self.value_col].agg(['mean', 'std'])
        self.mean_ = stats['mean']
        self.std_ = stats['std']
        self.global_mean_ = X[self.value_col].mean()
        self.global_std_ = X[self.value_col].std()
        return self

    def transform(self, X):
        X = X.copy()
        mean = X[self.group_col].map(self.mean_)
        std = X[self.group_col].map(self.std_)

        if self.fallback_to_global:
            mean = mean.fillna(self.global_mean_)
            std = std.fillna(self.global_std_)

        std = std.replace(0, np.nan)   # нулевой std даст inf

        X[f'{self.value_col}_std_{self.group_col}'] = std
        X[f'{self.value_col}_absz_{self.group_col}'] = (X[self.value_col] - mean) / std

        return X
    # def fit_transform(self,X,y= None):
    #     temp = self.fit(X,y)
    #     return self.transform(temp)

pipeline = sklearn.pipeline.Pipeline([('fill_cat', CategoricalFiller()),
                                      ('z_score, std', GroupStdFeatures()),
                                      ('model', CatBoostClassifier(
    iterations=3000,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU',
    cat_features = all_cat_columns
))
                                     ])
```
