## prepare data

```python
train_identity.drop('isFraud',axis = 1,inplace=True)
all_data = pd.merge(data,train_identity,how='left',on='TransactionID')
all_data.info()
```

```python
for col in all_data:
    if issubclass(all_data[col].dtype.type, str):
        if all_data[col].isna().sum() > 0:
            all_data[col] = all_data[col].fillna('miss')
```

```python
all_data = all_data.sort_values('TransactionDT')
```

```python
from sklearn.model_selection import train_test_split
X_train,y_train = all_data.drop('isFraud',axis = 1), all_data['isFraud']
X_train,X_val,y_train,y_val = train_test_split(X_train,y_train,test_size = 0.2,random_state = 42)
```

```python
all_cat_columns = []
for col in X_train.columns:
    if issubclass(X_train[col].dtype.type, str):
        all_cat_columns.append(col)
all_cat_columns
```

```python
y_train.mean(),y_val.mean()
```

```python
y_train.shape,y_val.shape
```

## boosting

```python
from catboost import CatBoostClassifier,Pool

train_pool = Pool(
    data=X_train,
    label=y_train,
    cat_features=all_cat_columns
)

val_pool = Pool(
    data=X_val,
    label=y_val,
    cat_features=all_cat_columns
)


model_pr_rec = CatBoostClassifier(
    iterations=1500,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU'
)
model_pr_rec.fit(train_pool, eval_set=(X_val, y_val))
```

```python
from sklearn.metrics import classification_report, roc_curve

y_pred_boost = model_pr_rec.predict(X_val)
y_pred_proba_boost = model_pr_rec.predict_proba(X_val)[:, 1]
print(classification_report(y_val,y_pred_boost))
```

```python
model_pr_rec_2 = CatBoostClassifier(
    iterations=2200,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU'
)
model_pr_rec_2.fit(train_pool, eval_set=(X_val, y_val))
```

```python
model_pr_rec_3 = CatBoostClassifier(
    iterations=3000,
    learning_rate=0.1,
    depth=6,
    eval_metric= 'PRAUC',
    custom_metric=['Precision','Recall'],
    class_weights=[1, 10],
    verbose=100,
    task_type='GPU'
)
model_pr_rec_3.fit(train_pool, eval_set=(X_val, y_val))
```

```python
y_pred_boost_3 = model_pr_rec_3.predict(X_val)
y_pred_proba_boost_3 = model_pr_rec_3.predict_proba(X_val)[:, 1]
print(classification_report(y_val,y_pred_boost_3))
```

```python
importances = model_pr_rec_3.get_feature_importance(prettified=True)
sns.barplot(y=importances.head(20)['Importances'], x=importances.head(20)['Feature Id']).tick_params(axis='x', rotation=45)
```

## random forest

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'), all_cat_columns)
    ],
    remainder='passthrough'
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_val_encoded = preprocessor.transform(X_val)
```

```python
X_train_encoded
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(300,criterion='entropy',max_depth= 100,n_jobs = -1,class_weight = {0:1,1:10})
rf.fit(X_train_encoded,y_train)
rf_preds_proba = rf.predict_proba(X_val_encoded)
y_pred_rf = rf.predict(X_val_encoded)
print(classification_report(y_val,y_pred_rf))
```

```python
X_train.info()
```

```python
feature_names = preprocessor.get_feature_names_out()
importances = pd.Series(rf.feature_importances_, index=feature_names).sort_values(ascending=False)
sns.barplot(y=importances.head(20), x=importances.head(20).index).tick_params(axis='x', rotation=45)
```

## comparison

```python
from sklearn.metrics import precision_recall_curve
pr_rf,rec_rf,_ = precision_recall_curve(y_val,rf_preds_proba[:,1])
pr_boost,rec_boost,_ = precision_recall_curve(y_val,y_pred_proba_boost_3)

fig,ax = plt.subplots()
ax.set_xlabel('recall')
ax.set_ylabel('precision')
ax.plot(rec_rf,pr_rf,label = 'random forest')
ax.plot(rec_boost,pr_boost,label = 'boosting')
ax.legend()
```
